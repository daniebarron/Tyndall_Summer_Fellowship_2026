```python
import numpy as np
from ase import io

from pfp_api_client.pfp.estimator import Estimator, EstimatorCalcMode

from matplotlib import pyplot as plt
from pfcc_extras.visualize.view import view_ngl
```

```python
trajs = []
for i in range(16):
    try:
        trajs.append(io.Trajectory(f'CF3/{i:0>2d}_MD-trajectory.traj'))
    except:
        break

probes = {'Energy [eV]': lambda atoms: [atoms.get_total_energy(), atoms.get_potential_energy()],
          'Temperature [K]': lambda atoms: atoms.get_temperature(),
          'Height F [Å]': lambda atoms: atoms.get_positions()[atoms.get_atomic_numbers() == 9, -1],
          'Height Si [Å]': lambda atoms: atoms.get_positions()[atoms.get_atomic_numbers() == 14, -1],
          'Height C [Å]': lambda atoms: atoms.get_positions()[atoms.get_atomic_numbers() == 6, -1],
          'Height H [Å]': lambda atoms: atoms.get_positions()[atoms.get_atomic_numbers() == 1, -1]
         }

fig, axs = plt.subplots(nrows=len(probes), figsize=(2*6.4, len(probes)*4.8), sharex=True)

time = [0]

for traj in trajs:
    time = time[-1] + .005 * np.arange(len(traj))

    results = {probe: [] for probe in probes}

    for i, atoms in enumerate(traj):
        for probe in probes:
            results[probe].append(probes[probe](atoms))

    for i, probe in enumerate(probes):
        axs[i].plot(time, results[probe])
        axs[i].set_prop_cycle(None)

for i, probe in enumerate(probes):
    axs[i].set_ylabel(probe)

view_ngl(trajs[-1])
```
